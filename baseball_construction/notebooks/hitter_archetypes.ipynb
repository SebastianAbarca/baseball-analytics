# Module 6 — Hitter Archetypes

Classification order: Complete Hitter → TTO → Spectrum (Pure Contact / Pure Power)

**Spectrum:** power_score / (power_score + contact_score) × 100  
0 = Arraez, 50 = balanced, 100 = Gallo

```python
import sys
sys.path.insert(0, "../modules")
from hitter_archetypes import (
    classify_primary, compute_spectrum, compute_free_swinger,
    compute_speed_modifier, compute_disruptiveness,
    build_hitter_profile, profiles_to_dataframe,
)
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.patches as mpatches
%matplotlib inline
```

```python
# Representative player archetypes
players = {
    "Trout (Complete)":  {"wRC_plus":76,"OBP":85,"ISO":88,"BB_pct":90,"ZContact_pct":72,"K_pct":55,"HR_FB_pct":82,"Barrel_pct":90,"EV_90_pct":95,"Contact_pct":60,"AVG_pct":75,"OBP_ISO_gap_pct":45},
    "Gallo (TTO)":       {"K_pct":90,"BB_pct":72,"ISO":82,"wRC_plus":55,"OBP":62,"ZContact_pct":30,"HR_FB_pct":95,"Barrel_pct":92,"EV_90_pct":98,"Contact_pct":15,"AVG_pct":20,"OBP_ISO_gap_pct":5},
    "Arraez (Contact)":  {"K_pct":5,"BB_pct":50,"ISO":15,"wRC_plus":68,"OBP":65,"ZContact_pct":45,"HR_FB_pct":8,"Barrel_pct":10,"EV_90_pct":28,"Contact_pct":97,"AVG_pct":98,"OBP_ISO_gap_pct":92},
    "Stanton (Power)":   {"K_pct":80,"BB_pct":55,"ISO":92,"wRC_plus":72,"OBP":58,"ZContact_pct":42,"HR_FB_pct":92,"Barrel_pct":88,"EV_90_pct":98,"Contact_pct":38,"AVG_pct":40,"OBP_ISO_gap_pct":12},
}
results = {name: classify_primary(m) for name, m in players.items()}
for name, r in results.items():
    spec = f"{r["spectrum_score"]:.1f}" if r["spectrum_score"] else "N/A"
    print(f"{name:22s}: {r["type_code"]} {r["type"]:22s} conf={r["display_confidence"]:3d}%  spectrum={spec}")
```

```python
# Visualise spectrum for a set of players
import numpy as np
names = list(players.keys())
spectra = [classify_primary(m)["spectrum_score"] or 50 for m in players.values()]
type_codes = [classify_primary(m)["type_code"] for m in players.values()]
color_map = {"T1":"gold","T2":"steelblue","T3":"seagreen","T4":"tomato"}
colors = [color_map.get(tc,"grey") for tc in type_codes]

fig, ax = plt.subplots(figsize=(10,3))
ax.barh(names, spectra, color=colors, height=0.5)
ax.axvline(50, color="black", linestyle="--", alpha=0.5, label="Balanced (50)")
ax.axvspan(45,55, alpha=0.08, color="grey", label="Grey zone")
ax.set_xlim(0,100)
ax.set_xlabel("Spectrum Score (0=Pure Contact, 100=Pure Power)")
ax.set_title("Hitter Spectrum by Archetype")
patches = [mpatches.Patch(color=v,label=k) for k,v in color_map.items()]
ax.legend(handles=patches, loc="lower right")
plt.tight_layout()
plt.show()
```
