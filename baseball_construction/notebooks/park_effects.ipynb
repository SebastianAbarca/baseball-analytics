# Module 4 — Park Effects

wOBA-based and HR-based park factors from Statcast PA outcomes.

**raw_PF = home_rate / road_rate** (neutral = 1.0)  
**regressed_PF = (2/3 × raw) + (1/3 × 1.0)**  
**pitcher_friendly** = inverted wOBA PF percentile (high = suppresses offense)

```python
import sys
sys.path.insert(0, "../modules")
import pandas as pd
import matplotlib.pyplot as plt
from ingest import pull_statcast_range
from park_effects import extract_pa_outcomes, build_park_profile
%matplotlib inline
```

```python
df = pull_statcast_range("2023-06-01", "2023-06-07", label="test")
pa = extract_pa_outcomes(df)
print(f"{len(pa):,} PA outcomes | {pa["is_hr"].sum()} HRs | league wOBA = {pa["woba_value"].mean():.3f}")
```

```python
# Full pipeline (3 teams meet 200-PA threshold in 1-week sample)
profile = build_park_profile(df, season=2023)
profile[["team","home_pa","road_pa","woba_pf_raw","woba_pf","hr_pf","pitcher_friendly","hr_pf_pct"]]
```

### Note on sample size
One week of data allows only 3 teams to reach the 200-PA floor. Run  to get all 30 teams with stable park factors.

```python
# Visualise park factors
if len(profile) > 0:
    fig, ax = plt.subplots(figsize=(8, 4))
    ax.barh(profile["team"], profile["woba_pf"] - 1, color=["steelblue" if x < 0 else "coral" for x in profile["woba_pf"] - 1])
    ax.axvline(0, color="black", linewidth=0.8)
    ax.set_xlabel("wOBA Park Factor vs Neutral (regressed)")
    ax.set_title("Park Factor by Team (3-team 1-week sample)")
    plt.tight_layout()
    plt.show()
```
